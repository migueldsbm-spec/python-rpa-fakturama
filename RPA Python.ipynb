## Desafio

- Vamos automatizar a extração de informações de um sistema e envio de um relatório por e-mail
- No nosso caso, para todo mundo conseguir fazer o mesmo programa, o nosso "sistema" vai ser o Gmail, mas o mesmo processo pode ser feito com qualquer programa do seu computador e qualquer sistema
    - Passo 1: Entrar no sistema (entrar no Gmail)
    - Passo 2: Entrar em uma aba específica do sistema onde tem o nosso relatório (Aba Contatos)
    - Passo 3: Exportar o Relatório (Exportar Contatos)
    - Passo 4: Pegar o relatório exportado, tratar e pegar as informações que queremos
    - Passo 5: Preencher/Atualizar informações do sistema (No nosso caso, criar um e-mail e enviar)

```python
import pyautogui
import time

pyautogui.alert(
    'O código vai começar. Não mexa em NADA enquanto o código estiver rodando. Quando finalizar eu te aviso'
)

pyautogui.PAUSE = 2

# abrir menu iniciar
pyautogui.press('win')

# abrir chrome
pyautogui.write('chrome')
pyautogui.press('enter')

time.sleep(2)

# acessar gmail
pyautogui.write('gmail.com')
pyautogui.press('enter')

# clicar para escrever o assunto para o destinatario e o assunto do email
pyautogui.moveTo(94, 223, duration=1)
pyautogui.click()
time.sleep(3)

# escrever destinatario
pyautogui.write('qualquer@gmail.com')
pyautogui.press('enter')

# ir para o assunto
pyautogui.press('tab')
pyautogui.write(
    'Relatorio Automatico.',
    interval=0.01
)

# ir para a mensagem
pyautogui.press('tab')
pyautogui.write(
    'Ola, este email foi enviado automaticamente utilizando Python.',
    interval=0.01
)

# enviar o email
pyautogui.moveTo(836, 699, duration=1)
pyautogui.click()

pyautogui.alert('Automação finalizada com sucesso. Fique a vontade para mexer.')
print('Fim do Código')
```
